```python
# Import libraries for data analysis and virtualization
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("../data/creditcard.csv")

# Quick preview of structure
df.head()
```

```python
# Understand dataset shape and column types
df.shape
df.info()
```

```python
# Statistical summary of numeric features
df.describe()
```

```python
# Distribution of transaction amounts
display(df["Amount"].describe())

# Visual check of distribution
df["Amount"].hist(bins=30)
plt.title("Transaction Amount Distribution")
plt.show()
```

```python
# Compare fraud vs normal transactions
df.groupby("Class")["Amount"].describe()
```

```python
# Convert time from seconds to hours
df["Time_hours"] = df["Time"] / 3600

# Visualize temporal distribution
df["Time_hours"].hist(bins=100)
plt.title("Transaction Time Distribution")
plt.show()
```

```python
# Separate distributions for fraud and normal transactions
df[df["Class"] == 0]["Time_hours"].hist(bins=100)
plt.title("Normal Transactions Time Distribution")
plt.show()

df[df["Class"] == 1]["Time_hours"].hist(bins=100)
plt.title("Fraud Transactions Time Distribution")
plt.show()
```

```python
# Log transform to reduce skewness
df["amount_log"] = np.log1p(df["Amount"])

# High-value transaction indicator (top 10%)
df["is_high_value"] = (df["amount_log"] > df["amount_log"].quantile(0.9)).astype(int)
```

```python
# Compare key engineered and raw features
df.groupby("Class")[["amount_log", "Time_hours", "is_high_value"]].agg(["mean", "std", "median"])
```
